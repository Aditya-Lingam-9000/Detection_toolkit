```python
from ultralytics import YOLO

model = YOLO('yolo11m.pt')
```

```python
results = model.train(
    data = '/content/dataset/data.yaml',
    imgsz = 640,
    epochs=50,
    plots=True,
    batch=16,
    workers=2,
)
```

```python
# Run detection on a new invoice image
# The model will use the 'best.pt' weights from your training run automatically
outputs = model.predict(source='/content/dataset/test/images/26-1_png.rf.67f83f9d20f0c116f294b352f1777f22.jpg', save=True, imgsz=640)
```
